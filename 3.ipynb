```python
tabla = []

for renglon in range(1, 11):
    fila = []

    for columna in range(1, 11):
        resultado = 0

        for i in range(columna):
            resultado = resultado + renglon

        fila.append(resultado)

    tabla.append(fila)


def imprimir_tabla(tabla):
    for fila in tabla:
        for numero in fila:
            print(numero, end="\t")
        print()

def consultar_producto(tabla, renglon, columna):
    return tabla[renglon - 1][columna - 1]


print("BIENVENIDO A LA TABLA DE PITÁGORAS")
print("----------------------------------")

imprimir_tabla(tabla)

print("----------------------------------")

renglon = int(input("Escribe el primer factor (1 al 10): "))

while renglon < 1 or renglon > 10:
    print("Número fuera de rango. Intenta de nuevo.")
    renglon = int(input("Escribe el primer factor (1 al 10): "))

columna = int(input("Escribe el segundo factor (1 al 10): "))

while columna < 1 or columna > 10:
    print("Número fuera de rango. Intenta de nuevo.")
    columna = int(input("Escribe el segundo factor (1 al 10): "))


resultado = consultar_producto(tabla, renglon, columna)

print("----------------------------------")
print("El resultado es:")
print(renglon, "x", columna, "=", resultado)
print("----------------------------------")
```
