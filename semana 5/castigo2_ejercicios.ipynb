```python
#5
setlist = (
    ("Intro", 95),
    ("Riff asesino", 245),
    ("Balada", 302),
    ("Solo", 128),
    ("Encore", 409)
)

total_segundos = 0

for cancion, duracion_seg in setlist:
    total_segundos += duracion_seg

    minutos = duracion_seg // 60
    segundos_restantes = duracion_seg % 60

    if segundos_restantes < 10:
        seg_str = f"0{segundos_restantes}"
    else:
        seg_str = f"{segundos_restantes}"

    print(f"{cancion}: {minutos}:{seg_str}")

minutos_totales = total_segundos // 60
segundos_totales_restantes = total_segundos % 60

if segundos_totales_restantes < 10:
    seg_total_str = f"0{segundos_totales_restantes}"
else:
    seg_total_str = f"{segundos_totales_restantes}"

print(f"Duración total: {minutos_totales}:{seg_total_str}")
```

```python
#4
temps = (8, 14, 22, 27, 31, 19, 11)

etiquetas = []
contador = {"frío": 0, "templado": 0, "calor": 0}


for t in temps:

    if t < 12:
        categoria = "frío"
    elif 12 <= t <= 24:
        categoria = "templado"
    else:
        categoria = "calor"

    etiquetas.append(categoria)
    contador[categoria] += 1

predominante = ""
max_conteo = -1

for categoria, cantidad in contador.items():
    if cantidad > max_conteo:
        max_conteo = cantidad
        predominante = categoria

print(etiquetas)
print(contador)
print(f"Predomina: {predominante}")
```

```python
#3
paradas = (
    ("Querétaro", 0),
    ("San Juan del Río", 52),
    ("Tequisquiapan", 74),
    ("Bernal", 125),
    ("Jalpan", 260)
)

max_distancia = -1
tramo_mas_largo = ""

for i in range(len(paradas) - 1):
    origen, km_origen = paradas[i]
    destino, km_destino = paradas[i + 1]

    distancia_tramo = km_destino - km_origen

    print(f"{origen} → {destino}: {distancia_tramo} km")

    if distancia_tramo > max_distancia:
        max_distancia = distancia_tramo
        tramo_mas_largo = f"{origen} → {destino}"

print(f"Tramo más largo: {tramo_mas_largo} ({max_distancia} km)")
```

```python
#2
votos = [("Computer", 48), ("Starflyer 59", 22), ("Chinse Football", 35), ("Part Time", 15), ("Fontaines D.C.", 40)]

banda_ganadora = ""
max_votos = -1

suma_total = 0

for banda, cantidad in votos:

    suma_total += cantidad

    if cantidad > max_votos:
        max_votos = cantidad
        banda_ganadora = banda

promedio = suma_total / len(votos)

print(f"Ganador: {banda_ganadora} con {max_votos} votos")
print(f"Promedio: {promedio:.1f}")
print("Arriba del promedio:")

for banda, cantidad in votos:
    if cantidad > promedio:
        print(f"- {banda} ({cantidad})")
```

```python
#1
dias = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")
km = (120, 85, 0, 210, 95, 0, 150)

total_km = 0
dias_rodados = 0
max_km = -1
idx_max = -1

for i in range(len(km)):
    distancia = km[i]

    if distancia > 0:
        total_km += distancia
        dias_rodados += 1

    if distancia > max_km:
        max_km = distancia
        idx_max = i

if dias_rodados > 0:
    promedio_exacto = total_km / dias_rodados
    promedio_rodado = int(promedio_exacto * 10 + 0.5) / 10
else:
    promedio_rodado = 0.0

resultado_tupla = (total_km, dias_rodados, promedio_rodado, idx_max)
total, dias_con_rodada, promedio, posicion_max = resultado_tupla

print(f"Total: {total} km")
print(f"Días con rodada: {dias_con_rodada}")
print(f"Promedio por día rodado: {promedio:.1f} km")
print(f"Día más largo: {dias[posicion_max]} ({km[posicion_max]} km)")
```
