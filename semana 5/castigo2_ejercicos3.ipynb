```python
#15
cupones = {"METAL10": 10, "MOTO20": 20, "ROCK15": 15}


def aplicar(total, codigo):

    codigo = codigo.strip()
    codigo = codigo.upper()

    try:
        descuento = cupones[codigo]

        nuevo_total = total - (total * descuento / 100)

        return nuevo_total

    except KeyError:
        print("Cupón no válido")
        return total


codigos = [" moto20 ", "GRATIS", "metal10"]

for codigo in codigos:

    resultado = aplicar(1200, codigo)

    print("'" + codigo + "' → $" + str(resultado))
```

```python
#14
sala = [
    ["X", "L", "L"],
    ["L", "X", "L"],
    ["X", "X", "L"]
]


def reservar(sala):

    while True:

        try:
            fila = int(input("Escribe la fila: "))
            columna = int(input("Escribe la columna: "))

            if sala[fila][columna] == "X":
                print("Ocupado, elige otro")

            else:
                sala[fila][columna] = "X"
                print("¡Reservado!")
                return True

        except ValueError:
            print("Escribe números")

        except IndexError:
            print("Ese asiento no existe")


reservar(sala)


for fila in sala:

    for asiento in fila:
        print(asiento, end=" ")

    print()
```

```python
#13
capturas = ["10", "7", "x", "", "9", "8.5", " 6 "]

validas = []
invalidas = 0

for dato in capturas:

    try:
        numero = int(dato)
        validas.append(numero)

    except ValueError:
        invalidas = invalidas + 1


suma = 0

for numero in validas:
    suma = suma + numero

promedio = suma / len(validas)

print("Válidas:", validas)
print("Inválidas:", invalidas)
print("Promedio:", round(promedio, 2))
```

```python
#12
def precio_boleto(edad):

    if edad < 12:
        return 150

    elif edad < 60:
        return 300

    else:
        return 180


total = 0

for i in range(3):

    while True:

        try:
            edad = int(input("Escribe tu edad: "))

            if edad < 0 or edad > 120:
                print("Edad fuera de rango")

            else:
                break

        except ValueError:
            print("Escribe la edad con número")

    precio = precio_boleto(edad)

    print("Boleto: $", precio)

    total = total + precio


print("Total a pagar: $", total)
```

```python
#11
def rendimiento():

    while True:

        try:
            km = float(input("Escribe los kilómetros recorridos: "))
            litros = float(input("Escribe los litros cargados: "))

            resultado = km / litros

            return round(resultado, 1)

        except ValueError:
            print("Escribe solo números")

        except ZeroDivisionError:
            print("Los litros no pueden ser 0")


km_litro = rendimiento()

print("Rendimiento:", km_litro, "km/l")

if km_litro >= 20:
    print("Excelente")

elif km_litro >= 15:
    print("Normal")

else:
    print("Revisa tu moto")
```
