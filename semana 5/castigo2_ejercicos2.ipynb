```python
#10
abrev = {
    "q": "que",
    "xq": "porque",
    "tmb": "también",
    "pq": "porque",
    "bn": "bien",
    "tqm": "te quiero mucho"
}


def expandir(mensaje):

    mensaje = mensaje.lower()

    palabras = mensaje.split()

    traduccion = []

    for palabra in palabras:

        nueva = abrev.get(palabra, palabra)

        traduccion.append(nueva)

    resultado = " ".join(traduccion)

    return resultado


print(expandir("Q onda tmb voy al concierto xq tocan bn"))
print(expandir("tqm Mamá"))
```

```python
#9
precios = {"playera": 350, "gorra": 220, "parche": 80, "sudadera": 650}

carrito = ["playera", "gorra", "playera", "parche", "parche", "parche"]

contador = {}

for producto in carrito:

    if producto in contador:
        contador[producto] = contador[producto] + 1

    else:
        contador[producto] = 1


subtotal = 0

for producto, cantidad in contador.items():

    importe = precios[producto] * cantidad
    subtotal = subtotal + importe

    print(cantidad, "x", producto, "= $", importe)


print("Subtotal: $", subtotal)

if subtotal > 800:
    descuento = subtotal * 0.10
else:
    descuento = 0

total = subtotal - descuento

print("Descuento:", descuento)
print("Total:", total)
```

```python
#8
nombres = ["Ana", "Alberto", "Beto", "Carla", "Brenda", "Andrea", "Ceci", "Diego"]

grupos = {}

for nombre in nombres:

    inicial = nombre[0]

    if inicial in grupos:
        grupos[inicial].append(nombre)

    else:
        grupos[inicial] = [nombre]


for inicial, lista in grupos.items():
    print(inicial, "(", len(lista), "):", lista)
```

```python
#7
notas = {
    "Ana": [90, 85, 78],
    "Luis": [60, 72, 65],
    "Mía": [100, 95, 98],
    "Beto": [70, 68, 74]
}

def promedio(lista):
    suma = 0

    for nota in lista:
        suma = suma + nota

    resultado = suma / len(lista)

    return round(resultado, 1)


promedios = {}

for alumno, calificaciones in notas.items():

    prom = promedio(calificaciones)
    promedios[alumno] = prom

    if prom >= 70:
        print(alumno, ":", prom, "- Aprobado")
    else:
        print(alumno, ":", prom, "- Reprobado")


mejor = ""

for alumno, prom in promedios.items():
    if mejor == "" or prom > promedios[mejor]:
        mejor = alumno

print("Mejor promedio:", mejor)
```

```python
#6
inventario = {"bujía": 10, "cadena": 2, "balata": 6, "filtro": 4}
pedidos = [("bujía", 3), ("cadena", 5), ("espejo", 1), ("balata", 6), ("filtro", 1)]

agotados = []

for pieza, cantidad in pedidos:

    if pieza not in inventario:
        print(pieza, ": no existe en el taller")

    elif inventario[pieza] < cantidad:
        print(pieza, ": stock insuficiente (hay", inventario[pieza], ")")

    else:
        inventario[pieza] = inventario[pieza] - cantidad
        print(pieza, ": surtido (", cantidad, ")")

for pieza, cantidad in inventario.items():
    if cantidad == 0:
        agotados.append(pieza)

print(inventario)
print("Agotados:", agotados)
```
