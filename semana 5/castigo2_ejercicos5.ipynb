```python
#25
playlist = [
    "Master of Puppets - Metallica - 8:35",
    "Holy Wars - Megadeth - 6:32",
    "Fade to Black - Metallica - 6:57",
    "The Trooper - Iron Maiden - 4:10",
    "Hangar 18 - Megadeth - 5:14",
    "Aces High - Iron Maiden - 4:31"
]


def a_segundos(tiempo):

    partes = tiempo.split(":")

    minutos = int(partes[0])
    segundos = int(partes[1])

    resultado = minutos * 60 + segundos

    return resultado


def formato(seg):

    minutos = seg // 60
    segundos = seg % 60

    return str(minutos) + ":" + str(segundos).zfill(2)


artistas = {}

for cancion in playlist:

    partes = cancion.split(" - ")

    nombre = partes[0]
    artista = partes[1]
    duracion = partes[2]

    segundos = a_segundos(duracion)

    if artista in artistas:
        artistas[artista] = artistas[artista] + segundos

    else:
        artistas[artista] = segundos


total = 0
mayor = 0
mas_suena = ""

for artista, segundos in artistas.items():

    print(artista + ":", formato(segundos))

    total = total + segundos

    if segundos > mayor:
        mayor = segundos
        mas_suena = artista


print("Más tiempo:", mas_suena)
print("Total:", formato(total))
```

```python
#24
nombres = ("Ana", "Luis", "Mía", "Beto")

cal = [
    [85, 90, 78, 92],
    [60, 75, 70, 68],
    [95, 98, 100, 97],
    [70, 65, 80, 75]
]


def promedio(lista):

    suma = 0

    for numero in lista:
        suma = suma + numero

    resultado = suma / len(lista)

    return resultado


promedios = []

for i in range(len(nombres)):

    resultado = promedio(cal[i])

    promedios.append(resultado)

    print(nombres[i] + ":", resultado)


for columna in range(4):

    parciales = []

    for fila in range(4):

        parciales.append(cal[fila][columna])

    resultado = promedio(parciales)

    print("Parcial", columna + 1, ":", resultado)


mayor = 0
mejor = ""

for i in range(len(nombres)):

    if promedios[i] > mayor:
        mayor = promedios[i]
        mejor = nombres[i]


print("Mejor alumno:", mejor)
```

```python
#23
saldos = {"Ana": 500, "Luis": 200, "Mía": 300}

movs = [
    "Ana:-150",
    "Luis:-250",
    "Mía:+100",
    "Pedro:-50",
    "Ana:-cien",
    "Mía:-380",
    "Luis:-200"
]

errores = 0

for movimiento in movs:

    partes = movimiento.split(":")

    nombre = partes[0]
    monto = partes[1]

    try:

        cantidad = int(monto)

        if nombre not in saldos:
            raise KeyError

        nuevo_saldo = saldos[nombre] + cantidad

        if nuevo_saldo < 0:
            print(nombre, ": saldo insuficiente")

        else:
            saldos[nombre] = nuevo_saldo

    except ValueError:
        print("Monto inválido en '" + movimiento + "'")
        errores = errores + 1

    except KeyError:
        print(nombre, "no tiene pulsera")
        errores = errores + 1


print(saldos)
print("Movimientos con error:", errores)
```

```python
#22
partidos = (
    ("Pumas", "Tigres", 2, 1),
    ("América", "Pumas", 0, 0),
    ("Tigres", "América", 3, 2),
    ("Pumas", "América", 1, 2),
    ("Tigres", "Pumas", 1, 1)
)

puntos = {}

for local, visitante, goles_local, goles_visitante in partidos:

    if local not in puntos:
        puntos[local] = 0

    if visitante not in puntos:
        puntos[visitante] = 0

    if goles_local > goles_visitante:
        puntos[local] = puntos[local] + 3

    elif goles_local < goles_visitante:
        puntos[visitante] = puntos[visitante] + 3

    else:
        puntos[local] = puntos[local] + 1
        puntos[visitante] = puntos[visitante] + 1


print(puntos)

campeon = ""

for equipo, cantidad in puntos.items():

    if campeon == "" or cantidad > puntos[campeon]:
        campeon = equipo


print("Campeón:", campeon, "con", puntos[campeon], "puntos")
```

```python
#21
autos = [
    ("ABC123", 9, 11),
    ("XYZ789", 8, 15),
    ("MOT650", 10, 11),
    ("QRO001", 13, 13)
]


def cobro(entrada, salida):

    horas = salida - entrada

    if horas <= 1:
        precio = 20

    else:
        precio = 20 + (horas - 1) * 15

    return precio


cobros = {}

for placa, entrada, salida in autos:

    horas = salida - entrada

    precio = cobro(entrada, salida)

    cobros[placa] = precio

    print(placa, ":", horas, "h → $", precio)


total = 0

for precio in cobros.values():
    total = total + precio

print("Total del día: $", total)
```
