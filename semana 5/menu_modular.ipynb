### Descripción del programa

Este programa contiene un menú interactivo con cinco opciones que permiten practicar diferentes herramientas de Python: tuplas, diccionarios, excepciones y strings.

Cada sección funciona mediante una función independiente. El menú se repite hasta que el usuario selecciona la opción de salir.

```python
#A
def seccion_tuplas():
    print("\n--- SECCIÓN DE TUPLAS ---")

    # Creamos una tupla con 5 números
    numeros = (8, 3, 12, 5, 10)

    # Mostramos el tercer elemento
    print("Tupla original:", numeros)
    print("Tercer elemento:", numeros[2])

    # Pedimos dos números al usuario
    num1 = int(input("Escribe un número: "))
    num2 = int(input("Escribe otro número: "))

    # Creamos una nueva tupla
    nueva_tupla = numeros + (num1, num2)

    print("Nueva tupla:", nueva_tupla)

    # Convertimos la nueva tupla en una lista
    lista = list(nueva_tupla)

    # Ordenamos la lista
    lista.sort()

    print("Lista ordenada:", lista)

    # Función para sumar los números
    def sumar_tupla(datos):
        suma = 0

        for numero in datos:
            suma = suma + numero

        return suma

    print("Suma de la nueva tupla:", sumar_tupla(nueva_tupla))
```

```python
#B

def seccion_diccionarios():
    print("\n--- SECCIÓN DE DICCIONARIOS ---")

    # Creamos un diccionario con 3 contactos
    contactos = {
        "Ana": "4421234567",
        "Luis": "4429876543",
        "Maria": "4425555555"
    }

    # Pedimos un nuevo contacto
    nombre = input("Escribe el nombre del nuevo contacto: ")
    telefono = input("Escribe su teléfono: ")

    # Agregamos el contacto
    contactos[nombre] = telefono

    # Mostramos solamente los nombres
    print("\nNombres de los contactos:")

    for nombre in contactos:
        print(nombre)

    # Función para buscar un teléfono
    def buscar_contacto(datos, nombre):
        if nombre in datos:
            return datos[nombre]
        else:
            return None

    # Pedimos el nombre que queremos buscar
    buscar = input("\n¿Qué contacto quieres buscar? ")

    telefono_encontrado = buscar_contacto(contactos, buscar)

    if telefono_encontrado != None:
        print("El teléfono de", buscar, "es:", telefono_encontrado)
    else:
        print(buscar, "no está en tus contactos")
```

```python
#c

def seccion_excepciones():
    print("\n--- SECCIÓN DE EXCEPCIONES ---")

    try:
        # Pedimos dos números enteros
        num1 = int(input("Escribe el primer número entero: "))
        num2 = int(input("Escribe el segundo número entero: "))

        # Mostramos la suma
        suma = num1 + num2
        print("La suma es:", suma)

        # Realizamos la división
        division = num1 / num2
        print("La división es:", division)

    except ValueError:
        print("Error: debes escribir números enteros y no dejar espacios vacíos.")

    except ZeroDivisionError:
        print("Error: no se puede dividir entre cero.")
```

```python
#d
def seccion_strings():
    print("\n--- SECCIÓN DE STRINGS ---")

    # Creamos un mensaje
    mensaje = "La musica transforma cada concierto"

    # Mostramos la longitud del mensaje
    print("Mensaje original:", mensaje)
    print("Longitud del mensaje:", len(mensaje))

    # Convertimos el mensaje a mayúsculas
    print("Mensaje en mayúsculas:", mensaje.upper())

    # Reemplazamos una palabra
    mensaje_nuevo = mensaje.replace("musica", "alegria")

    print("Mensaje con reemplazo:", mensaje_nuevo)

    # Función para contar las palabras
    def contar_palabras(texto):
        palabras = texto.split()
        return len(palabras)

    print("Número de palabras:", contar_palabras(mensaje))
```

```python
#E

def menu():
    opcion = ""

    while opcion != "5":
        print("\n========== MENÚ PRINCIPAL ==========")
        print("1) Tuplas")
        print("2) Diccionarios")
        print("3) Excepciones")
        print("4) Strings")
        print("5) Salir")
        print("====================================")

        opcion = input("Elige una opción: ")

        if opcion == "1":
            seccion_tuplas()

        elif opcion == "2":
            seccion_diccionarios()

        elif opcion == "3":
            seccion_excepciones()

        elif opcion == "4":
            seccion_strings()

        elif opcion == "5":
            print("Gracias por utilizar el programa. ¡Hasta pronto!")

        else:
            print("Opción inválida. Intenta nuevamente.")


# Iniciamos el programa
menu()
```
