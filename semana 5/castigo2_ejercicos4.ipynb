```python
#20
prohibidas = ["tonto", "menso", "baboso"]


def censurar(frase):

    palabras = frase.split()

    contador = 0

    for i in range(len(palabras)):

        if palabras[i].lower() in prohibidas:

            palabras[i] = "*" * len(palabras[i])

            contador = contador + 1

    resultado = " ".join(palabras)

    return resultado, contador


frases = [
    "no seas menso y pásame el código",
    "Tonto el que no haga commit",
    "todo bien profe"
]


for frase in frases:

    texto, cantidad = censurar(frase)

    print(texto, "(", cantidad, "censuradas)")
```

```python
#19
alumnos = ["Bruce Dickinson", "Brian Dickinson", "Ana Lopez", "Andrea Lopez", "Angel Lopez"]

usados = []


def usuario(nombre):

    partes = nombre.split()

    inicial = partes[0][0].lower()
    apellido = partes[1].lower()

    base = inicial + apellido

    return base


for nombre in alumnos:

    base = usuario(nombre)

    nuevo = base
    numero = 2

    while nuevo in usados:

        nuevo = base + str(numero)
        numero = numero + 1

    usados.append(nuevo)

    print(nombre, "→", nuevo)
```

```python
#18
frase = "el rock nunca va a morir en esta ciudad"


def vocales(palabra):

    contador = 0

    for letra in palabra:

        if letra in "aeiou":
            contador = contador + 1

    return contador


palabras = frase.split()

resultados = {}

for palabra in palabras:

    cantidad = vocales(palabra)

    resultados[palabra] = cantidad


print(resultados)

mayor = 0
palabra_mayor = ""

for palabra, cantidad in resultados.items():

    if cantidad > mayor:
        mayor = cantidad
        palabra_mayor = palabra


print("Más vocales:", palabra_mayor, "(", mayor, ")")
```

```python
#17
def es_palindromo(texto):

    texto = texto.lower()
    texto = texto.replace(" ", "")

    if texto == texto[::-1]:
        return True

    else:
        return False


frases = ("Anita lava la tina", "ABBA", "Metallica", "Oso")

for frase in frases:

    resultado = es_palindromo(frase)

    print(frase, ":", resultado)
```

```python
#16
def evaluar(pw):

    fallas = []

    if len(pw) < 8:
        fallas.append("mínimo 8 caracteres")

    numero = False
    mayuscula = False
    espacio = False

    for letra in pw:

        if letra.isdigit():
            numero = True

        if letra.isupper():
            mayuscula = True

        if letra == " ":
            espacio = True

    if numero == False:
        fallas.append("falta un número")

    if mayuscula == False:
        fallas.append("falta una mayúscula")

    if espacio == True:
        fallas.append("sin espacios")

    return fallas


contrasenas = ["metal", "Harley2008", "road glide 08", "KLR650moto"]

for pw in contrasenas:

    resultado = evaluar(pw)

    if len(resultado) == 0:
        print(pw, ": segura")
    else:
        print(pw, ":", resultado)
```
